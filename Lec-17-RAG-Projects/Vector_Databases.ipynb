# Sample text data


```python
texts = [
    "The quick brown fox jumps over the lazy dog.",
    "A journey of a thousand miles begins with a single step.",
    "To be or not to be, that is the question.",
    "All that glitters is not gold.",
    "The early bird catches the worm.",
    "Artificial intelligence is transforming industries.",
    "Natural language processing is a subfield of AI.",
    "Machine learning algorithms learn from data.",
    "Deep learning is a subset of machine learning."
]
```

```python
# Format the data into a list of dictionaries
text_data = [{"id": f"doc_{i}", "text": text} for i, text in enumerate(texts)]
text_data
```

# Generating Embeddings using BGE-M3

```python
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
# Initialize BGE-m3 using SentenceTransformers
model = SentenceTransformer('BAAI/bge-m3')
```

```python
# Generate Embeddings
embeddings = model.encode(texts).tolist()

# Attach embeddings back to the data dictionary
for i, item in enumerate(text_data):
    item["embedding"] = embeddings[i]
```

```python
text_data[0]
```

```python
# Generate a query embedding tailored to find the AI related sentences
query_text = "Data Science"
query_embedding = model.encode([query_text])[0].tolist()
vector_size = len(query_embedding)

print(f"Processed {len(text_data)} documents.")
print(f"Query Embedding dimension: {len(query_embedding)}")
```

## Eculidean Distance

The **Euclidean distance**, which measures the straight-line distance between two points in a multi-dimensional space.

**The Formula**

The distance $d$ between two vectors A and B, with $n$ dimensions, is calculated as:

$$d(A,B) = \sqrt{\sum_{i=1}^{n} (A_i - B_i)^2}$$


Here's a breakdown of what each part of the formula does:

1.  **$(A_i - B_i)^2$ : Squared Difference**
    This part calculates the difference between the points along each dimension `i` and then squares the result. Squaring is crucial because it ensures the value is always **positive** (distance cannot be negative) and gives greater weight to larger differences.

2.  **$\sum_{i=1}^{n}$ : Summation**
    This symbol means to sum up all the squared differences calculated in the previous step, from the first dimension ($i=1$) to the final dimension ($n$).

3.  **$\sqrt{\dots}$ : Square Root**
    Finally, taking the square root of the total sum reverses the initial squaring operation. This brings the final value back into the original unit of distance, giving you the actual straight-line distance between the two points.

```python
import math

def euclidean_distance(vector1, vector2):
    """
    Calculates the Euclidean distance between two vectors.
    """
    if len(vector1) != len(vector2):
        raise ValueError("Vectors must have the same dimension.")

    sum_sq_diff = 0
    for i in range(len(vector1)):
        sum_sq_diff += (vector1[i] - vector2[i]) ** 2

    distance = math.sqrt(sum_sq_diff)
    return distance

# Calculate the cosine similarity between the query embedding and the embedding of the most similar text
similarity = euclidean_distance(query_embedding, embeddings[0])
print(f"Euclidean Distance between query and the first text: {similarity}")
```

## Cosine Similarity

The **Cosine Similarity** between two vectors, A and B, measures the cosine of the angle between them. This value indicates their directional similarity, regardless of their magnitude. A value of 1 means they point in the exact same direction, 0 means they are perpendicular (orthogonal), and -1 means they are in opposite directions.

 **The Formula**

The formula is:

$$\text{Similarity}(A,B) = \cos(\theta) = \frac{A \cdot B}{\|A\| \cdot \|B\|} = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \cdot \sqrt{\sum_{i=1}^{n} B_i^2}}$$


Here's an explanation of each component:

1.  **$A \cdot B$ : The Dot Product**
    The numerator, $\sum_{i=1}^{n} A_i B_i$, is the **dot product** of vectors A and B. It's calculated by multiplying the corresponding components of the two vectors and then summing up all those products. A larger dot product suggests the vectors are pointing in a similar direction.

2.  **$\|A\|$ and $\|B\|$ : The L2 Norm (Magnitude)**
    The denominator consists of the L2 norm (or magnitude) of each vector. The magnitude, $\|A\| = \sqrt{\sum_{i=1}^{n} A_i^2}$, is the length of the vector, calculated by taking the square root of the sum of the squares of its components.

3.  **Normalization**
    The final division by the product of the magnitudes is a crucial step called **normalization**. It scales the dot product, effectively canceling out the influence of the vectors' lengths. This ensures the final score reflects only their directional relationship, not their size.

```python
import math

def cosine_similarity(vector1, vector2):
    """
    Calculates the cosine similarity between two vectors using base Python.
    """
    if len(vector1) != len(vector2):
        raise ValueError("Vectors must have the same dimension.")

    dot_product = sum(v1 * v2 for v1, v2 in zip(vector1, vector2))
    norm_vector1 = math.sqrt(sum(v**2 for v in vector1))
    norm_vector2 = math.sqrt(sum(v**2 for v in vector2))

    if norm_vector1 == 0 or norm_vector2 == 0:
        return 0  # Handle the case of zero vectors

    similarity = dot_product / (norm_vector1 * norm_vector2)
    return similarity

# Calculate the cosine similarity between the query embedding and the embedding of the most similar text
similarity = cosine_similarity(query_embedding, embeddings[0])
print(f"Cosine similarity between query and the first text: {similarity}")
```

## Dot or Inner Product


The **dot product** (also known as the scalar product) is an operation that takes two vectors and returns a single number (a scalar). It provides a simple way to determine the relationship between two vectors. There are two primary ways to define it.

**Algebraic Definition**

This formula calculates the dot product by multiplying the corresponding components of the two vectors and adding up the results.

$$A \cdot B = \sum_{i=1}^{n} A_i B_i = A_1B_1 + A_2B_2 + \dots + A_nB_n$$

**Geometric Definition**

This definition provides more intuition about what the dot product represents by relating it to the angle between the vectors.

$$A \cdot B = \|A\| \|B\| \cos(\theta)$$

Here's a breakdown of the components:
* **$\|A\|$** and **$\|B\|$**: These represent the **magnitudes** (or lengths) of vectors A and B.
* **$\cos(\theta)$**: This is the cosine of the angle **$\theta$** between the two vectors. It effectively measures their alignment:
    * If they point in the same direction, $\cos(\theta) = 1$.
    * If they are perpendicular, $\cos(\theta) = 0$.
    * If they point in opposite directions, $\cos(\theta) = -1$.

This second formula clearly shows that the dot product is influenced by both the **magnitudes** of the vectors and their **directional alignment**. In essence, it measures how much of one vector points in the direction of the other.

```python
def dot_product(vector1, vector2):
    """
    Calculates the dot product of two vectors using base Python.
    """
    if len(vector1) != len(vector2):
        raise ValueError("Vectors must have the same dimension.")

    result = 0
    for i in range(len(vector1)):
        result += vector1[i] * vector2[i]
    return result

# Calculate the dot product between the query embedding and the embedding of the most similar text
dot_product_example = dot_product(query_embedding, embeddings[0])
print(f"Dot product between query and the most similar text: {dot_product_example}")
```

# Qdrant

```python
from qdrant_client import QdrantClient
from qdrant_client.models import VectorParams, Distance, PointStruct

client = QdrantClient(":memory:") # Using RAG as server to create database. But it will be gone once restart the session
collection_name = "cosine_collection"

client.create_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(size=vector_size, distance=Distance.COSINE) # EUCLID or DOT
)
```

```python
points = [
    PointStruct(
        id=i,
        vector=item["embedding"],
        payload={"text": item["text"]}
    ) for i, item in enumerate(text_data)
]

client.upsert(collection_name=collection_name, points=points)

print(f"\n*** Top Results using COSINE ***")
search_result = client.query_points(
    collection_name=collection_name,
    query=query_embedding,
    limit=2
)

for result in search_result.points:
    print(f"Score: {result.score:.4f} | Text: {result.payload['text']}")
```

# ChromaDB

```python
import chromadb

chroma_client = chromadb.Client()
collection_name = "cosine_collection"

try:
    chroma_client.delete_collection(name=collection_name)
except Exception:
    pass

chroma_collection = chroma_client.create_collection(
    name=collection_name,
    metadata={"hnsw:space": "cosine"} # l2: euclidean, or ip: dot or inner product
)

chroma_collection.add(
    ids=[item["id"] for item in text_data],
    embeddings=[item["embedding"] for item in text_data],
    metadatas=[{"text": item["text"]} for item in text_data]
)

chroma_res = chroma_collection.query(
    query_embeddings=[query_embedding],
    n_results=2
)

print(f"\n*** Results using COSINE ***")
for i in range(len(chroma_res['ids'][0])):
    text = chroma_res['metadatas'][0][i]['text']
    distance = chroma_res['distances'][0][i]
    print(f"Rank {i+1} | Distance: {distance:.4f} | Text: {text}")
```
