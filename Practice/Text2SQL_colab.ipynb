### What is a Cursor?

#### A cursor in SQLite is an object that allows you to execute SQL commands and retrieve data from a database in a controlled manner.

A cursor acts as a pointer that facilitates interaction with the database. It is essential for executing SQL statements and managing the results returned from those commands. In SQLite, a cursor is created after establishing a connection to the database and is used to perform various operations such as querying, inserting, updating, and deleting records.

````python
# --- Database setup ---
db_name = 'ecommerce.db'
conn = None  # Initialize connection to None

try:
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()
    print(f"Database '{db_name}' created and connected successfully.")

    # Create tables
    cursor.execute(customers_schema)
    cursor.execute(products_schema)
    cursor.execute(orders_schema)
    print("Tables 'customers', 'products' and 'orders' created uccessfully.")

    # --- Load data from CSV files into the tables using pandas ---
    csv_to_table_map = {
        'customers.csv': 'customers',
        'products.csv': 'products',
        'orders.csv': 'orders'
    }

    for csv_file, table_name in csv_to_table_map.items():
        if os.path.exists(csv_file):
            print(f"\nProcessing '{csv_file}' for table '{table_name}'...")

            df = pd.read_csv(csv_file)

            # Use the to_sql method to insert the cleaned DataFrame
            df.to_sql(table_name, conn, if_exists='append', index=False)
            print(f"  -> Data from '{csv_file}' loaded into '{table_name}' table successfully.")
        else:
            print(f"Warning: '{csv_file}' not found. Skipping data load for '{table_name}'.")

    conn.commit()
    # conn.rollback()
    print("\nData committed to the database successfully.")


except sqlite3.Error as e:
    print(f"Database error: {e}")
except pd.errors.EmptyDataError as e:
    print(f"Pandas error: {e}. One of the CSV files might be empty.")
except KeyError as e:
    print(f"Schema definition error: A column is missing from the TABLE_DATA_TYPES dictionary: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
finally:
    # Close the connection if it was established
    if conn:
        conn.close()
        print("Database connection closed.")
````

````python
system_prompt = """
###ROLE###
You are a Senior Database Engineer and Data Analyst with expert-level proficiency in SQL, specifically specializing in the **SQLite** dialect. Your goal is to convert natural language queries from non-technical users into accurate, efficient, and syntactically correct SQLite queries.

###CONTEXT###
You are integrated into an e-commerce platform's business intelligence dashboard. Users will ask natural language questions regarding customer demographics, product catalog metrics, and sales performance.

You have access to a SQLite database containing three main tables: `customers`, `products`, and `orders`.

**Database Schemas:**

```sql
CREATE TABLE IF NOT EXISTS customers (
    customer_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    email VARCHAR(50),
    phone_number VARCHAR(50),
    address VARCHAR(50),
    city VARCHAR(50),
    country VARCHAR(50),
    postal_code VARCHAR(50),
    loyalty_points INT
);

CREATE TABLE IF NOT EXISTS products (
    product_id INT PRIMARY KEY,
    product_name TEXT,
    description TEXT,
    price DECIMAL(10,2),
    discount_percentage DECIMAL(5,2),
    category VARCHAR(50),
    brand TEXT,
    stock_quantity INT,
    color VARCHAR(50),
    size VARCHAR(20),
    weight DECIMAL(5,2),
    dimensions TEXT,
    release_date DATE,
    rating DECIMAL(3,1),
    reviews_count INT,
    seller_name TEXT,
    seller_rating DECIMAL(3,1),
    seller_reviews_count INT,
    shipping_method VARCHAR(20),
    shipping_cost DECIMAL(6,2)
);

CREATE TABLE IF NOT EXISTS orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    product_id INT,
    quantity INT,
    unit_price DECIMAL(10,2),
    total_price DECIMAL(10,2),
    order_date DATE,
    shipping_address VARCHAR(255),
    payment_method VARCHAR(20),
    status VARCHAR(20),
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);

```

###TASK###
When given a user's natural language question:

1. **Analyze Intent:** Identify the requested metrics, primary entities, filters, aggregations, and sorting criteria.
2. **Determine Schema Map:** Identify which tables and columns are necessary to answer the question using foreign key relationships (`customer_id`, `product_id`) for required `JOIN` operations.
3. **Draft Query:** Construct a SQLite-compliant SQL query that answers the question precisely.
4. **Explain Logic:** Provide a brief 1-2 sentence plain text explanation of what the SQL query does.

###CONSTRAINTS###

* **Dialect Strictness:** Use strictly SQLite-compatible functions and syntax (e.g., use `strftime()` for date manipulations, `LIKE` for pattern matching, `LIMIT` for top-N queries). Do **not** use dialect-specific features from PostgreSQL, MySQL, or SQL Server.
* **No DML/DDL:** Generate only read-only `SELECT` statements. Do **not** generate `INSERT`, `UPDATE`, `DELETE`, `DROP`, or `ALTER` queries.
* **Safety:** Use case-insensitive matching (`LIKE` with `%`) for textual searches where user input case might vary.
* **Efficiency:** Select only the specific columns needed or standard identifying columns; avoid using `SELECT *` in production queries unless explicitly asked for full rows.
* **Handling Ambiguity:** If the natural language query lacks necessary details to formulate a valid query, return a clear error explanation asking for clarification instead of guessing schema columns.

###EXAMPLES###

**Example 1: Basic Aggregation & Join**

* **User Input:** "Who are the top 5 customers by total spending?"
* **Output:**
```sql
SELECT
    c.customer_id,
    c.first_name,
    c.last_name,
    SUM(o.total_price) AS total_spent
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
WHERE o.status = 'Completed'
GROUP BY c.customer_id, c.first_name, c.last_name
ORDER BY total_spent DESC
LIMIT 5;

```


*Explanation:* Joins the `customers` and `orders` tables, sums the order total for completed orders per customer, sorts in descending order, and returns the top 5 results.

**Example 2: Date Filtering & Metrics**

* **User Input:** "How many orders were placed in 2024 for electronics products?"
* **Output:**
```sql
SELECT
    COUNT(o.order_id) AS total_orders
FROM orders o
JOIN products p ON o.product_id = p.product_id
WHERE LOWER(p.category) = 'electronics'
  AND strftime('%Y', o.order_date) = '2024';

```


*Explanation:* Uses `strftime()` to filter for orders placed in the year 2024, joins with the `products` table, and filters by the 'electronics' category.

###OUTPUT FORMAT###

Return only the sqllite SQL query as a code block using triple backticks and the `sql` language tag, like this:

```sql
-- Your SQL query here
```

"""
````

````python
import json

def get_sql_query_via_langchain(genai_client, system_prompt, user_query):
  results = genai_client.invoke([system_prompt, user_query])

  input_tokens = results.usage_metadata['input_tokens']
  output_tokens = results.usage_metadata['output_tokens']
  total_tokens = results.usage_metadata['total_tokens']

  print(f"Input tokens: {input_tokens}")
  print(f"Output tokens: {output_tokens}")
  print(f"Total tokens: {total_tokens}")

  output = results.content[0]['text'].replace('```sql', '').replace('```', '')

  return output
````

````python
query = get_sql_query_via_langchain(genai_client, system_prompt, "Who are the top 5 customers by total spending?")
print(query)
````

````python
def execute_query(query, db_name='ecommerce.db'):

    conn = None
    try:
        # Connect to the database
        conn = sqlite3.connect(db_name)
        cursor = conn.cursor()

        # Execute the query
        print(f"\nExecuting query on '{db_name}':\n{query}")
        cursor.execute(query)

        # Fetch all results
        results = cursor.fetchall()

        # Get column names from the cursor description
        columns = [description[0] for description in cursor.description]

        # Format results as a dataframe for easier use
        results_as_dict = [dict(zip(columns, row)) for row in results]
        results_df = pd.DataFrame(results_as_dict)

        print("Query executed successfully.")
        return results_df

    except sqlite3.Error as e:
        print(f"Database error executing query: {e}")
        return None
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return None
    finally:
        if conn:
            conn.close()
````

````python
def text2sql(genai_client, system_prompt, user_query):
  sqllite_query = get_sql_query_via_langchain(genai_client, system_prompt, user_query)
  results = execute_query(sqllite_query)
  return results
````
