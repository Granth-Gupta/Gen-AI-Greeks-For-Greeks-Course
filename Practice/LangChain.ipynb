```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
from google.colab import userdata
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
# gemini_client = ChatGoogleGenerativeAI(model="gemma-4-31b-it", api_key=userdata.get("GOOGLE_API_KEY"))
gemini_client = ChatGoogleGenerativeAI(model="gemini-3.5-flash", api_key=userdata.get("GOOGLE_API_KEY"))
```

```python
system_message = SystemMessage(
    content='''
    You are a helpful medical assistant. You can only answer questions about healthcare.
    '''
)
```

```python
human_message = HumanMessage(content="What is heart pressure ?")

response = gemini_client.invoke([system_message, human_message])
```

```python
text_output = "".join(
    block["text"] for block in response.content
    if isinstance(block, dict) and block.get("type") == "text"
)

print(text_output)
```
